# Paso 1. Importaciones y configuración

En este primer paso, importamos las librerías necesarias para desarrollar la calculadora hipotecaria interactiva. Utilizamos `ipywidgets` para crear los campos de entrada y el botón de cálculo, y `IPython.display` para mostrar la interfaz y actualizar los resultados.

También definimos un tipo de interés anual ilustrativo del 3,5 % para realizar las primeras pruebas. Este valor es únicamente un ejemplo y no representa una oferta real de una entidad bancaria.

```python
# PASO 1: IMPORTACIONES Y CONFIGURACIÓN

import ipywidgets as widgets
from IPython.display import display, clear_output

# Tipo de interés anual ilustrativo para realizar pruebas.
# No es una oferta real de ningún banco.
TIPO_INTERES_ANUAL = 3.5

print("Calculadora hipotecaria inicializada correctamente.")
print(f"Tipo de interés anual ilustrativo: {TIPO_INTERES_ANUAL}%")
```

# Paso 2. Función de cálculo de la hipoteca

En este paso, desarrollamos la función principal de la calculadora, denominada `calcular_hipoteca()`. Esta función recibe cuatro parámetros: el precio de la vivienda, la entrada inicial, el tipo de interés anual y el plazo de la hipoteca en años.

Primero, se validan los datos introducidos para evitar cálculos incorrectos. Después, se calcula el importe del préstamo, se convierte el tipo de interés anual en mensual y se estima la cuota mensual mediante la fórmula financiera correspondiente a un préstamo con cuotas mensuales constantes.

Por último, se calculan el importe total que se pagaría durante toda la hipoteca y los intereses totales. Los resultados se devuelven en un diccionario para que puedan utilizarse posteriormente en la interfaz.

```python
# PASO 2: FUNCIÓN DE CÁLCULO DE LA HIPOTECA

def calcular_hipoteca(
    precio_vivienda,
    entrada,
    tipo_interes_anual,
    años
):
    # 1. Validar los datos de entrada
    if precio_vivienda <= 0:
        raise ValueError("El precio de la vivienda debe ser positivo.")

    if entrada < 0 or entrada >= precio_vivienda:
        raise ValueError(
            "La entrada debe ser igual o superior a cero "
            "e inferior al precio de la vivienda."
        )

    if tipo_interes_anual < 0:
        raise ValueError("El tipo de interés no puede ser negativo.")

    if años <= 0:
        raise ValueError("El plazo de la hipoteca debe ser positivo.")

    # 2. Calcular el importe del préstamo
    importe_prestamo = precio_vivienda - entrada

    # 3. Convertir el interés anual en interés mensual
    interes_mensual = tipo_interes_anual / 100 / 12

    # 4. Calcular el número total de cuotas
    numero_cuotas = años * 12

    # 5. Calcular la cuota mensual
    if interes_mensual == 0:
        cuota_mensual = importe_prestamo / numero_cuotas
    else:
        cuota_mensual = (
            importe_prestamo
            * interes_mensual
            * (1 + interes_mensual) ** numero_cuotas
            / ((1 + interes_mensual) ** numero_cuotas - 1)
        )

    # 6. Calcular el importe total pagado y los intereses
    importe_total = cuota_mensual * numero_cuotas
    intereses_totales = importe_total - importe_prestamo

    # 7. Devolver los resultados
    return {
        "importe_prestamo": importe_prestamo,
        "cuota_mensual": cuota_mensual,
        "tipo_interes_anual": tipo_interes_anual,
        "años": años,
        "importe_total": importe_total,
        "intereses_totales": intereses_totales
    }
```

# Paso 3. Creación de la interfaz interactiva

En este paso, creamos una interfaz que permite al usuario realizar una simulación hipotecaria sin necesidad de modificar el código Python.

La interfaz incluye campos para introducir el precio de la vivienda, la entrada inicial y el tipo de interés anual, además de un control deslizante para seleccionar el plazo de la hipoteca.

Cuando el usuario pulsa el botón «Calcular hipoteca», el programa recoge los valores introducidos y los envía a la función `calcular_hipoteca()`. A continuación, muestra la cuota mensual estimada, el importe total pagado y los intereses totales.

Si los datos no son válidos, se muestra un mensaje de error en lugar de realizar el cálculo.

```python

# PASO 3: INTERFAZ INTERACTIVA

# Campo 1: Precio de la vivienda
precio_input = widgets.FloatText(
    value=250000,
    description="Precio vivienda (€):",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="350px")
)

# Campo 2: Entrada inicial
entrada_input = widgets.FloatText(
    value=50000,
    description="Entrada (€):",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="350px")
)

# Campo 3: Tipo de interés anual
interes_input = widgets.FloatText(
    value=TIPO_INTERES_ANUAL,
    description="Interés anual (%):",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="350px")
)

# Campo 4: Plazo de la hipoteca
plazo_input = widgets.IntSlider(
    value=30,
    min=5,
    max=40,
    step=5,
    description="Plazo (años):",
    style={"description_width": "initial"},
    continuous_update=False,
    layout=widgets.Layout(width="400px")
)

# Botón para calcular
boton_calcular = widgets.Button(
    description="Calcular hipoteca",
    button_style="success",
    icon="calculator",
    layout=widgets.Layout(width="220px")
)

# Área donde se mostrarán los resultados
salida = widgets.Output()


# Función que se ejecuta al pulsar el botón
def al_calcular(boton):

    with salida:
        clear_output(wait=True)

        try:
            # Obtener los valores introducidos por el usuario
            precio = precio_input.value
            entrada = entrada_input.value
            interes = interes_input.value
            plazo = plazo_input.value

            # Ejecutar la función de cálculo
            resultado = calcular_hipoteca(
                precio_vivienda=precio,
                entrada=entrada,
                tipo_interes_anual=interes,
                años=plazo
            )

            # Mostrar los resultados
            print("SIMULACIÓN HIPOTECARIA")
            print("=" * 35)
            print(f"Precio de la vivienda: {precio:,.2f} €")
            print(f"Entrada: {entrada:,.2f} €")
            print(
                f"Importe del préstamo: "
                f"{resultado['importe_prestamo']:,.2f} €"
            )
            print(f"Tipo de interés anual: {interes:.2f} %")
            print(f"Plazo: {plazo} años")
            print("-" * 35)
            print(
                f"CUOTA MENSUAL ESTIMADA: "
                f"{resultado['cuota_mensual']:,.2f} €"
            )
            print(
                f"Importe total pagado: "
                f"{resultado['importe_total']:,.2f} €"
            )
            print(
                f"Intereses totales: "
                f"{resultado['intereses_totales']:,.2f} €"
            )
            print()
            print(
                "Nota: Estimación con interés fijo y cuotas mensuales "
                "constantes. No incluye impuestos, comisiones, seguros "
                "ni otros gastos."
            )

        except (ValueError, TypeError, OverflowError) as error:
            print(f"No se puede calcular la hipoteca: {error}")


# Conectar el botón con la función
boton_calcular.on_click(al_calcular)


# Mostrar la interfaz completa
display(
    widgets.VBox([
        widgets.HTML("<h2>Calculadora hipotecaria</h2>"),
        widgets.HTML(
            "<p>Introduce los datos para calcular tu cuota mensual.</p>"
        ),
        precio_input,
        entrada_input,
        interes_input,
        plazo_input,
        boton_calcular,
        salida
    ])
)
```
